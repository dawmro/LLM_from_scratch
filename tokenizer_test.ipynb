### 3. Create tokenizer class

```python
class SimpleTokenizerV1:
    def __init__(self, vocab: dict):
        # Direct assignment of the dict containing vocabulary mapping form string to int
        self.str_to_int = vocab
        # Reverse mapping
        self.int_to_str = {i:s for s,i in vocab.items()}

    def encode(self, text: str) -> int:
        # split text on whitespaces and spececial characters
        preprocessed = re.split(r'([,.;:?!"()\'_-]|\s)', text)
        # get rid of whitespaces
        preprocessed = [item.strip() for item in preprocessed if item.strip()]
        # assign token ID for each token
        ids = [self.str_to_int[s] for s in preprocessed]
        return ids
    
    def decode(self, ids: int) -> str:
        # convert token IDs into tokens and join individual tokens together
        text = " ".join([self.int_to_str[i] for i in ids])
        # replace spaces before specified characters
        text = re.sub(r'\s+([,.?!"()\'])', r'\1', text)
        return text
```

```python
# use tokenizer on input text
tokenizer = SimpleTokenizerV1(vocab=vocab)
text = "I'm a wizard and you are not. Ha!"
ids = tokenizer.encode(text)
print(ids)
```

```python
out = tokenizer.decode(ids)
print(out)
```

### 4. Modify tokenizer to handle unknown tokens

```python
all_tokens = sorted(list(set(preprocessed)))
all_tokens.extend(["<|endoftext|>", "<|unk|>"])

vocab = {token:integer for integer,token in enumerate(all_tokens)}
print(len(vocab.items()))
```

```python
class SimpleTokenizerV2:
    def __init__(self, vocab: dict):
        # Direct assignment of the dict containing vocabulary mapping form string to int
        self.str_to_int = vocab
        # Reverse mapping
        self.int_to_str = {i:s for s,i in vocab.items()}

    def encode(self, text: str) -> int:
        # split text on whitespaces and spececial characters
        preprocessed = re.split(r'([,.;:?!"()\'_-]|\s)', text)
        # get rid of whitespaces
        preprocessed = [item.strip() for item in preprocessed if item.strip()]
        # if entry not in vocab assign unknown token to it
        preprocessed = [
            item if item in self.str_to_int 
            else "<|unk|>" for item in preprocessed 
        ]
        # assign token ID for each token
        ids = [self.str_to_int[s] for s in preprocessed]
        return ids
    
    def decode(self, ids: int) -> str:
        # convert token IDs into tokens and join individual tokens together
        text = " ".join([self.int_to_str[i] for i in ids])
        # replace spaces before specified characters
        text = re.sub(r'\s+([,.?!"()\'])', r'\1', text)
        return text
```

```python
tokenizer = SimpleTokenizerV2(vocab)
text1 = "I'm a wizard and you are not. Ha!"
text2 = "I like apple juice - I drink it"
text = " <|endoftext|> ".join((text1, text2))

print(text)
```

```python
enc = tokenizer.encode(text)
print(enc)
```

```python
dec = tokenizer.decode(enc)
print(dec)
```

### 7. Data Loader

```python
# implementing data loader that iterates over the input dataset and returns the inputs and targets as PyTorch tensors
from torch.utils.data import Dataset, DataLoader
import torch



class GPTDatasetV1(Dataset):
    def __init__(self, txt, tokenizer, max_length, stride):
        self.input_ids = []
        self.target_ids = []

        # Tokenize the entire text
        token_ids = tokenizer.encode(txt, allowed_special={"<|endoftext|>"})

        # Use a sliding window to chunk the book into overlapping sequences of max_length
        for i in range(0, len(token_ids) - max_length, stride):
            input_chunk = token_ids[i:i + max_length]
            target_chunk = token_ids[i + 1: i + max_length + 1]
            self.input_ids.append(torch.tensor(input_chunk))
            self.target_ids.append(torch.tensor(target_chunk))

    def __len__(self):
        return len(self.input_ids)

    def __getitem__(self, idx):
        return self.input_ids[idx], self.target_ids[idx]



```

```python
def create_dataloader_v1(txt, batch_size=4, max_length=256, 
                         stride=128, shuffle=True, drop_last=True,
                         num_workers=0):

    # Initialize the tokenizer
    tokenizer = tiktoken.get_encoding("gpt2")

    # Create dataset
    dataset = GPTDatasetV1(txt, tokenizer, max_length, stride)

    # Create dataloader
    dataloader = DataLoader(
        dataset,
        batch_size=batch_size,
        shuffle=shuffle,
        drop_last=drop_last,
        num_workers=num_workers
    )

    return dataloader
```

```python
# test the dataloader with a batch size of 1 for an LLM with a context size of 4
text_file = "wizard_of_oz.txt"

#load sequence
with open(text_file, "r", encoding="utf-8") as f:
    raw_text = f.read()


print("PyTorch version:", torch.__version__)
dataloader = create_dataloader_v1(
    raw_text, batch_size=1, max_length=4, stride=1, shuffle=False
)

data_iter = iter(dataloader)
first_batch = next(data_iter)
print(first_batch)
```

```python
# illustrate the meaning of stride=1, let's fetch another batch from this dataset:
second_batch = next(data_iter)
print(second_batch)
```

```python
# use the data loader to sample with a batch size greater than 1
dataloader = create_dataloader_v1(raw_text, batch_size=8, max_length=4, stride=4, shuffle=False)

data_iter = iter(dataloader)
inputs, targets = next(data_iter)
print("Inputs:\n", inputs)
print("\nTargets:\n", targets)

# Note that we increase the stride to 4. This is to utilize the data set fully (we don't skip a single word) 
# but also avoid any overlap between the batches, since more overlap could lead to increased overfitting.
```
